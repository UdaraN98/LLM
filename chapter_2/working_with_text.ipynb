```python
# punctuation marks split

result = re.split(r'([,.]|\s)', text)
print(result)
```

```python
# expanding the function for more punctuation types

text = "Hello, world. Is this-- a test?"
result = re.split(r'([,.:;?_!"()\']|--|\s)', text)
result = [item.strip() for item in result if item.strip()]
print(result)
```

### 2.3 Converting tokens into token IDs

```python
# Creating a list of all unique tokens and sort them in alphabetical order

all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
print(vocab_size)
```

```python
# Creating the vocabulary

vocab = {
    token: integer for integer, token in enumerate(all_words)
}

for i, item in enumerate(vocab.items()):
    print(item)
    if i>= 50:
        break
```

### 2.3 Implementing a simple text tokenizer

```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        '''
        1. Stores the vocabulary as a class attribute for access in the encode and decode methods
        
        '''
        self.str_to_int = vocab
        self.int_to_str = {i:s for s, i in vocab.items()} # Creates in inverse vocab for the decoding task

    def encode(self, text):
        '''
        Method to convert the input text in to list of token ids by initially conveting the text into tokens and mapping the tokens
        to uniqe ids using the vocab
        '''
        # Tokenizing using our simple tokenization mechanism
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        # mapping the id using our vocab 
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    
    def decode(self, ids):
        '''
        Method to convert a list of token IDs back to tokens and ultimately into text
        '''
        # get the list of related tokens
        text = " ".join([self.int_to_str[i] for i in ids])

        # remove the whitespaces before punctuation marks

        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text

    
```

```python
# Let's tokenize and encode "The Verdict"

tokenizer = SimpleTokenizerV1(vocab)
text = """"It's the last he painted, you know,"
Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
```
